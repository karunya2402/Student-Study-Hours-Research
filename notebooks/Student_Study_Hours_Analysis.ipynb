#Student Study Hours vs Exam Performance

**Task 0- Loading Dataset Using Pandas**

```python
import pandas as pd

df = pd.read_csv("student_study_data.csv")

print(df)
```

**Task 1- Dataset Verification**

```python
print("Dataset Shape:", df.shape)
print("\nColumn Names:", df.columns.tolist())
print("\nMissing Values:")
print(df.isnull().sum())
```

**Task 2- Checking Data Types**

```python
print(df.dtypes)
```

**Task 3- Descripitive Statistical Analysis**

```python
print(df.describe())
```

**Task 4- Data Visualization Using Scatter Plot**

```python
import matplotlib.pyplot as plt

plt.scatter(df["Study_Hours"], df["Exam_Score"])

plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

plt.title("Study Hours vs Exam Performance")

plt.show()
```

**Task 5-Correlation Analysis**

```python
correlation = df["Study_Hours"].corr(df["Exam_Score"])

print("Correlation Coefficient:", correlation)
```

Interpretation:
+1 → Perfect positive correlation
0 → No linear correlation
-1 → Perfect negative correlation

**Task 6– Statistical Analysis Using Pearson Correlation Test**

```python
from scipy.stats import pearsonr

r, p_value = pearsonr(df["Study_Hours"], df["Exam_Score"])

print("Correlation Coefficient:", r)
print("P-value:", p_value)
```

Interpretation:
If p-value < 0.05 → Statistically significant relationship.
If p-value ≥ 0.05 → Not statistically significant.
